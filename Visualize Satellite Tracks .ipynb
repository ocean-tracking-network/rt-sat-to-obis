# Visualize Satellite Tracks on Kepler.gl

## Two options for Kepler viewer:

### Option 1. JPhub hosted: 

- In a command terminal: run `ssh -L 8000:localhost:8000  -i <your_ssl_pk_file> <your_user>@jphub.oceantrack.org`
- In web browser: http://localhost:8000/
Details see https://gitlab.oceantrack.org/otn-partner-nodes/jphub/-/wikis/Host-and-Use-Kepler.gl-Viewer-on-JPhub

###  Option 2. Public site  https://kepler.gl/demo





## 1. Upload data file e.g. `<cid>_tracks_<yyyymmdd>.csv` to Kepler viewer




## 2. Click `Add Layer`: under `Basic` select `Point`; under `Color based on` select `tag_req`



## 3. In Filters tab, click `Add Filter` select `date_time`



## 4 Adjust rolling window and click the play button
