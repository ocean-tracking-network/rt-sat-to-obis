## Manually Run ArgosQC for SMRU Tags - ArgosQC.smru_qc

1. Install required R package(s).
2. Run smru_qc for specified configuration file.

### (Run One Time) Install ArgosQC package 

Pinned to a stable verion. Contact the OTN data team for version upgrade.

```r
pinned_version <- "0.9.16"

options(install.packages.check.source = "no")
# Unload package prior to reinstall/update
unloadNamespace("ArgosQC")
remotes::install_github("ianjonsen/ArgosQC", ref = pinned_version, force = TRUE, INSTALL_opts = "--no-multiarch", dependency=NA)
```
