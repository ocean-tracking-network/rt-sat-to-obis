## Manually Run ArgosQC for SMRU Tags - ArgosQC.smru_qc

1. Install required R package(s).
2. Run smru_qc for specified configuration file.

### Install ArgosQC package - run this everytime to upgrade to the latest

#### Prerequisite:  R 4.5.2 and RTools 4.5 

```r
options(install.packages.check.source = "no")
options(pkgType = "binary")

# Unload package prior to reinstall/update
unloadNamespace("ArgosQC")
remotes::install_github("ianjonsen/ArgosQC@smru_qc-support-windows-2", force = TRUE, INSTALL_opts = "--no-multiarch")

# Load package and check version
library(ArgosQC)
packageDescription("ArgosQC", fields = "Version")
```
