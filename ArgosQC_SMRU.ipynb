```r
# Simple function to read your MDB file
read_smru_data <- function(cid = "ct182", 
                           mdb_dir = "D:/otn_gitlab/otn_github/rt-sat-to-obis/qc/mdb",
                           mdbtools_path = "C:/Users/hpz2/mdbtools-app/") {
  
  file_path <- file.path(mdb_dir, paste0(cid, ".mdb"))
  
  if(!file.exists(file_path)) {
    stop("File not found: ", file_path)
  }
  
  # Get list of tables
  cmd_tables <- paste0(shQuote(paste0(mdbtools_path, "mdb-tables")), " -1 ", shQuote(file_path))
  tables <- system(cmd_tables, intern = TRUE)
  tables <- tables[nchar(trimws(tables)) > 0]
  
  cat("Found tables:", paste(tables, collapse = ", "), "\n")
  
  # Read each table into a list
  smru <- list()
  for(table in tables) {
    # Export table to CSV format
    cmd_export <- paste0(shQuote(paste0(mdbtools_path, "mdb-export")), " ",
                        shQuote(file_path), " ", shQuote(table))
    
    # Read the exported data
    data_text <- system(cmd_export, intern = TRUE)
    
    if(length(data_text) > 1) {  # Has header + at least one row
      # Parse CSV
      tc <- textConnection(data_text)
      smru[[table]] <- read.csv(tc, stringsAsFactors = FALSE)
      close(tc)
      cat("  Read", table, ":", nrow(smru[[table]]), "rows\n")
    } else {
      cat("  Table", table, "is empty\n")
      smru[[table]] <- data.frame()
    }
  }
  
  return(smru)
}
```
