# Load NRT Results to DB

1. Connect to NRT Database 
2. Show uploaded 
2. Search for SSM results for specified program
3. Upload SSM resluts to NRT DB.

```python
from py_nrt.common import get_engine
from py_nrt import load_nrt_results as lnr
import itables
qc_output_path = 'qc'
```

## User input required

```python
engine = get_engine(r"py_nrt/database_conn_string.auth")
```

## Check OTN NRT backend DB and schema connectivity

```python
lnr.check_otn_nrt_backend(engine)
```

## Show Deployments by Campain

```python
db_nrt_metadata_df = lnr.show_db_deployments(engine)
```

## Show Alive Tags ( Last Update < 24 hours)

```python
lnr.show_db_nrt_status(engine)
```

```python
# Some programs: otn, imos, iraq
program_campaigns = lnr.get_qced_program_campaigns(qc_output_path)
programs = list(program_campaigns.keys())
ssmoutput_last_modified_map = lnr.get_campaign_qc_results_for_program(qc_output_path, programs[0])
ssmoutput_last_modified_map
```

## Load SSM output into NRT backend

```python
summary_df, _ = lnr.load_to_nrt_db(engine, ssmoutput_last_modified_map)
if summary_df is not None and not summary_df.empty:
    print('Load summary by tag.')
    itables.show(summary_df)
```
