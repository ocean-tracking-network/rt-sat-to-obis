# WildlifeComputers Project Config


1. Connects to WildlifeComputers portal (requires access_key and secret_key pair)
2. Shows a list of collaborator (tag owner) who shared projects with OTN.
3. Downloads 

```python
from py_nrt import argos_wc_project_config as awpc
import itables
```

## Fill in Access Key and Secret Key for OTN account

```python
# Ask OTN DC for the keys.
access_key = ""
secret_key = ""

# Destination folder for wc_qc config files and tag downloads.
dest_path = r'input'

# Set to True to see API payload and response.
verbose=False
```

## Output Collaborator ID and email Pairs

```python
# Get collaborators
collaborators_df, collaborator_id_widget = awpc.wc_get_collab_ids(a_key=access_key, s_key=secret_key, verbose=verbose)
```

## Show Deployment(s) by Selected Collaborator

```python
deployment_df = awpc.get_deployments_for_owner_id(
    a_key=access_key, 
    s_key=secret_key, 
    owner_id=collaborator_id_widget.value, 
    verbose=verbose
)
```

## (Optional) Download Selected Tag for Observation

```python
awpc.download_tag(
    a_key=access_key, 
    s_key=secret_key, 
    deployment_df=deployment_df, 
    dest_path = dest_path, 
    verbose=verbose
)
```

## Generate WC_QC Configuration Files

```python
# Default settings
program = 'irap'
time_step = 3

# Project specific settings
project_id = 'vmlppws'
common_name = 'grey seal'
species = 'Halichoerus grypus'
release_site = 'Kouchibouguac National Park'
state_country = 'Canada'

awpc.create_wc_qc_config(
    dest_path=dest_path,
    a_key=access_key, 
    s_key=secret_key, 
    owner_id=collaborator_id_widget.value,
    program = program,
    project_id = project_id,
    time_step=time_step,
    common_name=common_name,
    species=species,
    release_site=release_site,
    state_country=state_country
)
```
